### មុខងារ​អនាមិក (Lambda)
- ការពិពណ៌នា: Lambda គឺជា​មុខងារ​តូច ដែលគ្មានឈ្មោះ ប្រើពាក្យគន្លឹះ `lambda`។
- សញ្ញាសម្ព័ន្ធ: lambda arguments: expression
- ការប្រើប្រាស់: ប្រើសម្រាប់មុខងារសង្ខេប ដែលត្រូវបានប្រើម្តងឬពីរី ក្នុងកន្លែងដែលត្រូវការមុខងារ។

#### លក្ខណៈពិសេសនៃ Lambda
1. ព្យញ្ជនៈតែមួយ: Lambda ចំណាត់ថ្នាក់មានសមីការប៉ុន្មានតែមួយ។
2. គ្មានឈ្មោះ: វាមិនត្រូវការឈ្មោះទេ ប៉ុន្តែអាចចាត់តាំងឲ្យទៅនាមអថេរ។
3. ប្រើក្នុងបន្ទាត់: ជាញឹកញាប់ប្រើនៅក្នុង `map()`, `filter()`, ឬ `sorted()`។

```python
square = lambda x: x ** 2
print(square(5))
```

```python
add = lambda a, b: a + b
print(add(10, 15))
```

```python
def multiplier(n):
    return lambda x: x * n

times_three = multiplier(4)
print(times_three(20))
```

#### Map? (ផែនការ map)

```python
nums = [1, 2, 3, 4,6]
squares = map(lambda x: x * x,nums )
#print(squares)
print(list(squares))
```

#### Filter? (ជម្រើស filter)

```python
nums = [10, 15, 20, 25, 30]
even_nums = filter(lambda x: x % 2 == 0, nums)
print(list(even_nums))
```

#### sort? (តម្រៀប)

```python
people = [("Alice", 25,"Siem Reap"), 
          ("Devid", 20,"Phnom Penh"), 
          ("Charlie", 30,"Kandal")]
print(people)
sorted_people = sorted(people, key=lambda person: person[0])
print(sorted_people)
```

### អាគុយម៉ង់តាមទីតាំង (Positional Arguments)

### អាគុយម៉ង់ចម្លង (Arbitrary Arguments: *args និង **kwargs)
- ប្រើសម្រាប់ចាត់ចែងចំនួនអាគុយម៉ង់ដែលមិនច្បាស់។
- បង្កើតភាពបត់បែនក្នុងការរចនាមុខងារ។

1. *args: អាគុយម៉ង់តាមទីតាំងច្រើន
- ការពិពណ៌នា: *args បង្គោលអាគុយម៉ង់បន្ថែម។
- ប្រភេទទិន្នន័យ: ត្រូវបានរក្សាទុកជា tuple.
- ករណីប្រើ: ប្រើពេលដែលចង់ដោះស្រាយច្រើនអាគុយម៉ង់ជាក្រុម។

```python
def add(a,b,c):
    return a+b+c
print(add(3,4,5))
```

```python
def add_plus(a,b,*args):
    return a+b+sum(args)
```

```python
print(add_plus(1, 4,5,5))
```

```python
def hello(class_name, *args):
    if args:
        for arg in args:
            print(f"welcome {arg} to class {class_name}")
    else:
        print(f"hello {class_name}")
```

```python
hello("B201","Hoeb Surneang","Dany","Rity","Reak Smey","ratana")
```

```python
hello("Hoeb Surneang")
```

2. **kwargs: អាគុយម៉ង់ដោយឈ្មោះច្រើន
- ការពិពណ៌នា: **kwargs បង្គោលអាគុយម៉ង់ដែលមានឈ្មោះបន្ថែម។
- ប្រភេទទិន្នន័យ: ត្រូវបានរក្សាទុកជា dictionary.
- ករណីប្រើ: ប្រើពេលដែលចង់ដោះស្រាយអាគុយម៉ង់ដែលមានឈ្មោះ។

```python
def show_info(name,age,**kwargs):
    print(f"Name:{name}, Age:{age}")
    print(f"Detail Information")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
show_info("Alice", 25) 
```

```python
show_info("Alice", 25,school="BBU",love="\tcoding",like="music") 
```

#### កម្រិត ៧ ការហៅឡើងវិញ (Recursion)

```python
def factorial(n):
    if n==0:
        return 1
    else:
        print(f"factorial({n})= {n} * {n-1} = {n*(n-1)}")
        return n*factorial(n-1)
```

```python
factorial(6)
```

#### កម្រិត ៩ តុបតែងមុខងារ (Function Decorators)

```python
def my_decorator(func):
    def wrappers():
        print("Befor fuction code excecute " )
        func()
        print("Befor fuction code excecute " )
    return wrappers
```

```python
@my_decorator
def greet():
    print("Hello!")
```

```python
greet()
```

#### កម្រិត ១០ មុខងារដ៏ខ្ពស់ (Advanced Functions)

```python
def square(x):
    return x ** 2
```

```python
result= apply(square,5)
print(result)
```
